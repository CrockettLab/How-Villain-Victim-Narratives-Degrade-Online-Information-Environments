# Behavioral Experiment Engagement Analyses
## Load Libraries

```python
# for data handling
import pandas as pd
import numpy as np

# to run regressions on differentially private data
from svinfer_ss1.linear_model import LinearRegression
from svinfer_ss1.processor import AthenaProcessor, DataFrameProcessor

# to run linear mixed effects models
import statsmodels.formula.api as smf
```

## Load Data

```python
df = pd.read_csv('headline_level_engagement_data.csv')
df.head()
```

## Prepare Data

```python
# subset on clear headlines
df_clear = df[df['clarity_majority'] == 1]
```

```python
# code the dyad variables
df_clear = df_clear.copy()
df_clear["villain_binary"] = df_clear["villain_presence_majority"].astype(int)
df_clear["victim_binary"]  = df_clear["victim_presence_majority"].astype(int)
df_clear["dyad_binary"] = (
    (df_clear["villain_binary"] == 1) & (df_clear["victim_binary"] == 1)
).astype(int)
```

## Analyses
**1. Headlines containing components of the dyad evoke more outrage than those without.**
### Dyad ~ Outrage

```python
m_dyad = smf.mixedlm(
    "angers ~ dyad_binary",
    data = df_clear,
    groups = df_clear["parent_domain"],
    re_formula = "~1",
    vc_formula = {"url": "0 + C(url_rid)"},
).fit(reml=True)
```

```python
m_dyad.summary()
```

### Villains ~ Outrage

```python
m_villain = smf.mixedlm(
    "angers ~ villain_binary",
    data = df_clear,
    groups = df_clear["parent_domain"],
    re_formula = "~1",
    vc_formula = {"url": "0 + C(url_rid)"},
).fit(reml=True)
```

```python
m_villain.summary()
```

### Victims ~ Outrage

```python
m_victim = smf.mixedlm(
    "angers ~ victim_binary",
    data = df_clear,
    groups = df_clear["parent_domain"],
    re_formula = "~1",
    vc_formula = {"url": "0 + C(url_rid)"},
).fit(reml=True)
```

```python
m_victim.summary()
```

**2. Outrage is more strongly associated with increased shares for headlines containing components of the dyad than those without.**

```python
# create subsets for the dyad, villains, and victims
dyad_df = df_clear[df_clear['dyad_binary'] == 1]
no_dyad_df = df_clear[df_clear['dyad_binary'] == 0]

villains_df = df_clear[df_clear['villain_binary'] == 1]
no_villains_df = df_clear[df_clear['villain_binary'] == 0]

victims_df = df_clear[df_clear['victim_binary'] == 1]
no_victims_df = df_clear[df_clear['victim_binary'] == 0]
```

```python
x_s2=[10.0]
```

### Shares ~ Angers (Dyad)

```python
dyad_data = DataFrameProcessor(dyad_df)
model = LinearRegression(
    ["angers",], 
    "shares",
    x_s2,
).fit(dyad_data)

print(f"beta_tilde is: \n{model.beta}")
print(f"beta_tilde's standard error is: \n{model.beta_standarderror}")
print(f"beta_tile's variance-covariance matrix: \n{model.beta_vcov}")
print(f"estimated residual variance is {model.sigma_sq}")
```

```python
no_dyad_data = DataFrameProcessor(no_dyad_df)
model = LinearRegression(
    ["angers",], 
    "shares",
    x_s2,
).fit(no_dyad_data)

print(f"beta_tilde is: \n{model.beta}")
print(f"beta_tilde's standard error is: \n{model.beta_standarderror}")
print(f"beta_tile's variance-covariance matrix: \n{model.beta_vcov}")
print(f"estimated residual variance is {model.sigma_sq}")
```
